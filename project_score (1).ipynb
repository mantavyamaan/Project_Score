## Situational Modifiers:

These are continuous multipliers — not binary gates — that capture real-world conditions which systematically affect project execution but do not fit cleanly into any single parameter. They adjust the final score up or down based on contextual factors.

**M1** — Data completeness confidence. Measures how much of the scoring data is based on confirmed information versus estimates, assumptions, or default values.

M1 = 0.5 + 0.5 * d_c

Where d_c is the proportion of all variables across all parameters that are populated from confirmed data rather than estimates, ranging from 0 to 1. The floor of 0.5 ensures that a project with entirely estimated data is not zeroed out, but it is significantly penalised.

**M2** — Organisational change risk. Measures whether the client is undergoing a merger, acquisition, leadership change, reorganisation, or strategic pivot that could destabilise the project.

1.00 — Stable organisation with no known changes

0.85 — Minor restructuring or leadership transition in a non-relevant department

0.70 — Major leadership or strategy change in progress that directly affects the project stakeholders

**M3** — Contract and commercial complexity. Measures how difficult the legal and commercial terms are to execute against.

1.00 — Standard terms and template SOW

0.85 — Custom terms required, some negotiation needed

0.70 — Complex legal negotiation expected, such as international or multi-entity arrangements

0.60 — Extreme contractual complexity: multiple jurisdictions, novel IP structures, heavy penalty clauses

**M4** — Currency and payment jurisdiction risk. Measures exposure to exchange rate fluctuations, cross-border payment delays, or clients in jurisdictions with payment reliability issues.

1.00 — Domestic, stable currency, reliable payment infrastructure

0.85 — International but stable market and well-established payment channels

0.70 — Emerging market, volatile currency, or delayed payment norms

**M5** — Execution complexity multiplier. Captures logistical difficulty: multiple time zones, language barriers, multi-location delivery, complex approval chains, or distributed stakeholder groups.

1.00 — Single-vertical, single-location, single-timezone, straightforward delivery

0.85 — Two verticals or two time zones

0.70 — Multi-vertical, multi-timezone

0.60 — Multi-vertical, multi-location, multi-language, complex stakeholder matrix

**M6** — Seasonal and cyclical pressure. Measures whether the project coincides with high-demand periods that compress timelines or compete for resources.

1.00 — Slow season: capacity is available, winning work is valuable, slight score boost

0.95 — Normal period
