## 1. Scope Definition & Requirements Clarity

This parameter measures how well-defined the project's boundaries, deliverables, and acceptance criteria are. Poorly defined scope is the single largest predictor of project failure-it causes rework, timeline slippage, budget overruns, and client dissatisfaction. The variables I will be using are:

**Rd** — Requirements documentation completeness. The fraction of total expected requirements that have been formally written, reviewed, and agreed upon by both parties. [0,1]

**Sa** — Scope agreement formality. Measures the contractual rigour of the scope lock-in. [0,1]

0.25: Verbal agreement only

0.50: Email confirmation

0.75: Signed scope document

1.00: Signed SOW with formal change control process

**Sc** — Number of scope changes after kickoff. Every scope change after work begins introduces re-planning overhead, potential rework, communication friction, and margin erosion. [integer]

**Uc** — Use-case specificity. How concretely the client has described the scenarios, workflows, and user journeys the deliverable must support. [0,1]

**Ac** — Acceptance criteria defined. Whether there exist explicit, testable, measurable conditions that define when each deliverable is considered "done." [0,1]

**Am** — Ambiguity index. The proportion of open questions, undefined edge cases, contradictory requirements, or assumptions still unresolved. Lower is better. [0,1]

**Dp** — Deliverables explicitly listed. Whether every output artefact (file, page, component, report, campaign, test result) is named, described, and quantified [0,1]

**Cr** — Client responsiveness to clarifications. How quickly and thoroughly the client resolves open questions when asked during scoping. [0,1]

**Vt** — Number of project verticals touched (design, development, digital marketing, AI, PTaaS, branding). Multi-vertical projects have inherently higher scope complexity and cross-functional coordination overhead. [integer,1–6]

**Coh** — Cross-vertical scope coherence. When multiple verticals are involved, this measures how well-integrated and non-contradictory the requirements are across them. [0,1]

**Rev** — Revision and iteration policy clarity. Whether revision limits, feedback rounds, and the process for handling change requests are contractually defined.
[0,1]

0: No revision limits discussed

0.5: Verbal understanding

1.0: Revision limits contractually defined with change-order process

**Fc** — Feature creep risk (inverse scale — lower is worse). Estimated likelihood the client will expand scope mid-project without renegotiation. [0,1]

0: Near-certain scope creep (vague requirements, "we'll figure it out" attitude)

0.5: Moderate risk

1.0: Locked scope, formal change-order process agreed

**Ref** — Reference material provided. Whether the client has supplied concrete reference inputs (mood boards, brand guides, competitor examples, existing assets, API documentation, content). [0,1]

0: No references

0.5: Verbal descriptions or loose mood references

1.0: Comprehensive reference package provided


**Vd** — Vertical-specific requirement depth. Measures how thoroughly the requirements address vertical-specific concerns. [0,1]

For design: brand guidelines, colour palettes, typography specifications, responsive breakpoints.

For development: tech stack defined, API documentation, hosting details, performance benchmarks.

For marketing: target audience defined, KPIs specified, analytics access, channel strategy.

For AI: data availability confirmed, model expectations documented, integration endpoints defined.

For PTaaS: asset scope defined, compliance requirements listed, reporting format agreed.

For branding: market research provided, positioning brief written, competitive analysis shared

<br>

F_scope (scope foundation) = 0.25* rd + 0.2* sa + 0.2* ac + 0.2*dp + 0.15* rev

C_scope (scope clarity) = 0.3* ac+ 0.3*(1-am)+ 0.2*vd + 0.2* ref

S_stab (scope stability) = max(0, 1 - (ln(1+sc))/ln(7)) * (0.5 + 0.5*Fc)

S_s = 0.4* F_scope+ 0.3* C_scope+ 0.3* S_stab

V_mod (cross-vertical modifier) = 1- 0.05*(vt-1)*(1-coh)

K_cli (client cooperation counter) = 0.5 + 0.5* cr

<br>

**X1** = 0.5 * S_s + 0.3 * V_mod + 0.2* K_cli

## 2. Revenue, Budget & Financial Health

This parameter unifies pre-sales financial evaluation (is the project worth pursuing?) with in-flight financial tracking (is the project staying on budget?). It measures margin viability, payment security, budget adherence, cost recovery, and revenue significance. The variables I will be using are:

**V** — Estimated total project value (contract amount), it is what the client pays (revenue).

**C** — Estimated total cost to deliver, thorized for spending on the project. It represents the financial limit approved by management, finance, or the client.

**Tb** — Total approved project budget, it is the amount that has been formally authorized for spending on the project. It represents the financial limit approved by management, finance, or the client.

**Bs** — Budget spent to date.

**Pp** — Project percentage physically complete (actual measured progress, not elapsed time). [0,1]

**Ps** — Payment structure security. [0,1]

0.20: Net-90 (90 days after the invoice had been issued) or longer, no milestone payments

0.40: Net-60, partial upfront

0.60: Net-30, 30% upfront

0.80: 50% upfront, milestone-based payments

1.00: 100% upfront or escrow-secured

**Bc** — Budget confirmation level. [0,1]

0: No budget discussion has occurred

0.25: Ballpark range mentioned verbally

0.50: Verbal confirmation of specific budget

0.75: Written confirmation via email or letter

1.00: Signed SOW or purchase order issued

**Crr** — Cost recovery rate: total invoiced divided by total spent. [0,1]

**Co** — Number of cost overruns flagged during execution. [integer]

**Cm** — Client payment compliance: on-time payments divided by total invoices sent. [0,1]

**Rv** — Revenue visibility: fraction of total project revenue that is contracted versus at-risk. [0,1]

**Cp** — Change-order premium captured: fraction of scope changes for which additional fees were charged. [0,1]

**Mg** — Current gross margin (actual, measured during execution). [0,1]

**Rr** — Revenue recurrence probability: likelihood this project leads to recurring revenue (retainer, subscription, maintenance contract). [0,1]

**Dk** — Currency and payment jurisdiction risk. [0,1]

1.0: Domestic, stable currency, reliable payment infrastructure

0.7: International but stable market and currency

0.5: Emerging market, volatile currency, or delayed payment norms

<br>

Mr (margin ratio) = max ( 0, min( (V-C)/V, 1) )

V_n (normalised projected value) = 1+ 1/( e^(-0.00003)*(V-150000)). This centres at ₹1.5L and scales smoothly

CPI (cost performance index) = (pp*tb)/(bs+ $\epsilon$)
where $\epsilon$= 10^(-9) when bs=0

CPI_n (normalised) = min(CPI,1.3)/1.3

CPI > 1 means under budget; CPI < 1 means over budget. Capped at 1.3 because extraordinary under-spending may indicate unstarted work rather than efficiency

H_cash (cash health) = (2*min(crr, 1) *cm)/ (min(crr,1) + cm + $\epsilon$ )

O_pen (overrrun penalty)= max(0, ln(1+co)/ln(5))

M_stab (margin and revenue stability) = 0.3 * min (Mg/0.4, 1) + 0.25*rv + 0.25*cp + 0.25*rr

F_pre (pre-sales finance confidence) = 0.3*mr + 0.25*ps + 0.2*bc+ 0.15*v_n + 0.1*dk

F_flight (in-flight financial health) = 0.25*CPI_n + 0.25*H_cash + 0.25*O_pen + 0.25*M_stab

For projects that have not yet started (bs=0, pp=0), the in-flight component is not yet meaningful. The composite adapts:

α=min(Pp*2,1)

**X2** = (1-α)* F_pre + α*( 0.4*F_pre + 0.6* F_flight)


```python
v= data[1]["v"]
c= data[1]["c"]
tb= data[1]["tb"]
bs= data[1]["bs"]
pp= data[1]["pp"]
ps= data[1]["ps"]
bc= data[1]["bc"]
crr= data[1]["crr"]
co= data[1]["co"]
cm= data[1]["cm"]
rv= data[1]["rv"]
cp= data[1]["cp"]
mg= data[1]["mg"]
rr= data[1]["rr"]
dk= data[1]["dk"]
cp= data[1]["cp"]
mg= data[1]["mg"]
rr= data[1]["rr"]
dk= data[1]["dk"]

mr= max(0, min((v-c)/v, 1))
v_n= 1+ 1/(math.exp(-0.00003)*(v-150000))
cpi = pp*tb/(bs+1e-9)
cpi_n= min(cpi,1.3)/1.3
h_cash= (2*min(crr,1)*cm)/(min(crr,1)+cm+1e-9)
o_pen= max(0, math.log(1+co)/math.log(5))
m_stab= 0.3*min(mg/0.4,1)+0.25*rv+0.25*cp+0.25*rr
f_pre= 0.3*mr + 0.25*ps + 0.2*bc + 0.15*v_n + 0.1*dk
f_flight= 0.25*cpi_n + 0.25*h_cash + 0.25*o_pen + 0.25*m_stab

a=min(pp*2,1)
x2= (1-a)*f_pre+a*(0.4*f_pre+0.6*f_flight)

print(x2)
```

## 3. Resource Allocation, Team Competency & Availability

This parameter measures whether the right people, in the right quantity, with the right skills, at the right seniority level, are assigned to the project — and whether they have the bandwidth, tooling, and resilience to deliver without burnout. The variables I will be using are:

**Ta** — Team availability ratio: allocated hours divided by required hours. [0,1]

**Se** — Average seniority-experience score of the assigned team members. [0,1]

**Dm** — Domain match: the team's specific experience in the project's vertical (design, development, AI, PTaaS, etc.). [0,1]

**Sk** — Skill coverage: proportion of all required skills that are available in-house without subcontracting. [0,1]

**Kc** — Key-person concentration risk. Measures how dependent the project is on a single individual. 1 = total single-point-of-failure; 0 = fully distributed. Lower is better. [0,1]

**Bs** — Bench strength: availability of qualified backup resources who could step in. [0,1]

**Tl** — Tool and license readiness: whether all required software, environments, subscriptions, and licenses are provisioned and accessible. [0,1]

**Ct** — Cross-training level within the team: can team members cover each other's roles in case of absence? [0,1]

**Rl** — Number of roles currently filled. [integer]

**Rr** — Number of roles required for full staffing. [integer]

**Ot** — Current overtime ratio of the assigned team. Lower is better. [0,1]

**On** — Number of other active projects competing for the same team member. [integer]

**Sr** — Seniority match: whether the available team members match the seniority level the project demands. [0,1]

0: Only juniors available for a senior-level project

0.5: Mixed coverage, some seniority gaps

1.0: Appropriate seniority levels fully available

**Od** — Outsource dependency: proportion of the total work that must be outsourced or subcontracted. [0,1]

**Ld** — Lead/PM availability. [0,1]

0: No project manager available

0.5: PM available but overloaded across multiple projects

1.0: Dedicated PM can be assigned

**Bn** — Burnout risk: current team workload stress level measured through overtime trends, leave patterns, and self-reported fatigue. [0,1]

**Pp** — Portfolio precedent: number of similar projects the team has completed previously. [integer]

<br>

Ta' = min(1,Ta)

S_fill = min(rl/rr , 1)

S_cap = ( 2 * S_fill * Ta' ) / S_fill + Ta' + $\epsilon$

The harmonic mean ensures that if either headcount or hours is deficient, the composite drops significantly. Having all roles filled but zero availability is just as bad as having all the availability but missing half the roles.

C_comp (competency composite) = 0.25*se + 0.25*dm + 0.2*sk + 0.15*sr + 0.15*ct

E_conf (experience confidence) = min( ln(1+pp)/ln(11) , 1)

R_res (resilience factor) = (1 - 0.4 * kc * (1-bs)) * (1-0.5*od)

When key-person concentration is high and bench strength is low, resilience suffers. Outsource dependency further degrades resilience because we have less control over quality, availability, and communication.

F_fat (fatigue and contention ratio) = e^(-1.5*Ot) * max(0, 1 - On/8) * (1 - 0.4*Bn)

I_r = 0.5*Tl + 0.5*Ld

**X3**= 0.25*S_cap + 0.2*C_comp + 0.1*E_conf + 0.15*R_res + 0.15*F_fat + 0.15* I_r

```python
ta= data[2]["ta"]
se= data[2]["se"]
dm= data[2]["dm"]
sk= data[2]["sk"]
kc= data[2]["kc"]
bs= data[2]["bs"]
tl= data[2]["tl"]
ct= data[2]["ct"]
rl= data[2]["rl"]
rr= data[2]["rr"]
ot= data[2]["ot"]
on= data[2]["on"]
sr= data[2]["sr"]
od= data[2]["od"]
ld= data[2]["ld"]
bn= data[2]["bn"]
pp= data[2]["pp"]

ta_prime= min(1,ta)
s_fill= min(rl/rr,1)
s_cap= (2*s_fill*ta_prime)/(s_fill+ta_prime+1e-9)

c_comp= 0.25*se + 0.25*dm + 0.2*sk + 0.15*sr + 0.15*ct
e_conf= min(math.log(1+pp)/math.log(11),1)
r_res= (1-0.4*kc*(1-bs))*(1-0.5*od)

f_fat= math.exp(-1.5*ot)*max(0,1-on/8)*(1-0.4*bn)
i_r= 0.5*tl+0.5*ld

x3= 0.25*s_cap + 0.2*c_comp + 0.1*e_conf + 0.15*r_res + 0.15*f_fat + 0.15*i_r

print(x3)
```

## 4. Timeline, Milestone Health & Delivery Realism

This parameter measures schedule performance from both a pre-commitment and in-flight perspective: whether the proposed timeline is realistic, whether milestones are being hit, and whether the remaining schedule is achievable given current velocity. Variables I will be using are:

**Tr** — Client's requested timeline in days. [integer]

**Te_est** — Estimated realistic timeline in days (our team's honest assessment). [integer]

**Td** — Total approved project duration in days (final agreed timeline). [integer]

**Te** — Days elapsed since project kickoff. [integer]

**Mc** — Milestones completed on or before their target date. [integer]

**Mt** — Total milestones planned. [integer]

**Md** — Milestones that were delayed (missed their target date). [integer]

**Ad** — Average delay in days per delayed milestone. [float]

**Bf** — Buffer days remaining in the schedule. [integer]

**Cd** — Number of critical-path delays (delays on tasks that directly push the project end date). [integer]

**Vl** — Velocity: current sprint or phase throughput divided by planned throughput. [0,1]

**Vl'** = min(1,Vl)

**Bl** — Backlog health: percentage of remaining work items that are estimated and prioritised. [0,1]

**De** — Number of external dependencies that could block progress (client approvals, third-party APIs, content delivery, legal review). [integer]

**Ph** — Phased delivery possible. [0,1]

0: Must deliver everything at once (big-bang delivery)

0.5: Some phasing possible

1.0: Fully phased or iterative delivery model agreed

**Pn** — Penalty and SLA risk. [0,1]

0: No contractual penalties for delay

0.5: Soft penalties or reputational risk

1.0: Hard contractual penalties for missed deadlines

**Tz** — Timezone overlap with client= |Z_client-Z_team|

Z_client = GMT + x

Z_client = GMT + Y

|Z_client-Z_team|= Absolute timezone difference in hours

<br>

tz_prime= max(0, 1- (12-tz)/12)

T_p (timeline pressure ratio) = min ( Tr/Te_est, 1)

SPI (schedule performance ratio) = Mc / ( Mc + Md + $\epsilon$ )

D_sev (delay severity) = e^(-0.5*Cd) * max(0, 1 - Ad/(Bf+1))

Elapsed ratio = Te / Td+$\epsilon$

Completion Ratio = Mc / Mt+$\epsilon$

P_sched (schedule pressure index) = 1 / ( 1 + e^(-8*(Completion Ratio-Elapsed Ratio)))

V_health (velocity health) = 0.6*(min(1.2 , Vl)/1.2) + 0.4* Bl

D_struct = 0.3*max(0, 1 - De/8 ) + 0.3*Ph + 0.2*(1 - 0.4*Pn) + 0.2*Tz

X4_pre = 0.35*T_p + 0.30*D_struct + 0.35*V_health

X4_flight = 0.25 * SPI + 0.2 * D_sev + 0.2 * P_sched + 0.15 * V_health + 0.1*T_p + 0.1 * D_struct

α = min ((Te/(Td+$\epsilon$))*2, 1)

**X4** = (1-α)*X4_pre + α * X4_flight



```python
tr= data[3]["tr"]
te_est= data[3]["te_est"]
td= data[3]["td"]
te= data[3]["te"]
mc= data[3]["mc"]
mt= data[3]["mt"]
md= data[3]["md"]
ad= data[3]["ad"]
bf= data[3]["bf"]
cd= data[3]["cd"]
vl= data[3]["vl"]
vl_prime= min(1,vl)
bl= data[3]["bl"]
de= data[3]["de"]
ph= data[3]["ph"]
pn= data[3]["pn"]
tz= data[3]["tz"]

tz_prime= max(0,1-abs(12-tz)/12)
t_p= min(tr/te_est,1)
spi= mc/(mc+md+1e-9)
d_sev= math.exp(-0.5*cd)*max(0,1-ad/(bf+1))
elapsed_ratio= te/td+1e-9
completion_ratio= mc/mt+1e-9

p_sched= 1/(1+math.exp(-8*(completion_ratio-elapsed_ratio)))

v_health= 0.6*(min(1.2,vl)/1.2)+0.4*bl

d_struct= 0.3*max(0,1-de/8)+0.3*ph+0.2*(1-0.4*pn)+0.2*tz_prime

x4_pre= 0.35*t_p+0.30*d_struct+0.35*v_health
x4_flight= 0.25*spi+0.2*d_sev+0.2*p_sched+0.15*v_health+0.1*t_p+0.1*d_struct

a= min((te/(td+1e-9))*2,1)

x4= (1-a)*x4_pre+a*x4_flight

print(x4)
```

## 5. Client Profile, Reliability & Communication

This parameter evaluates the client as a partner — their trustworthiness, professionalism, communication patterns, accessibility, payment history, and how constructively they engage throughout the project lifecycle. Vatiables I will be using are:

**Ch** — Client history with our company. [0,1]

0: New client, no prior interaction

0.5: One past project or informal relationship

1.0: Multiple successful projects together

**Cr** — Client reputation in the market, based on publicly available information, industry reviews, and professional network signals. [0,1]

**Rt** — Average client response time in hours. [float]

**Fd** — Feedback depth: how detailed, specific, and actionable client feedback is (versus vague or contradictory direction). [0,1]

**Dma** — Decision-maker accessibility. [0,1]

0: Cannot reach the decision-maker; dealing exclusively with intermediaries who lack authority

0.5: Occasional access to the decision-maker

1.0: Direct, regular access to the final decision-maker

**Py** — Payment history (for returning clients) or creditworthiness signal (for new clients). [0,1]

0: Known late payer, disputed invoices, or poor credit signals

0.5: Unknown or new client (neutral)

1.0: Consistently pays on time, clean financial record

**Coc** — Communication clarity: how clearly the client articulates needs, provides feedback, makes decisions, and manages their side of the relationship. [0,1]

**Sp** — Single point of contact consistency: does the client maintain a consistent, empowered contact person, or do different people give conflicting direction? [0,1]

**Ec** — Number of distinct communication channels actively used (email, Slack, calls, video, in-person, project management tool). [integer]

**Dc** — Total documented communications (meeting notes with decisions, confirmed email threads, signed-off documents). [integer]

**Pp** — Proactive participation: the client volunteers information, assets, approvals, and context without being asked or chased. [0,1]

**Ng** — Negative engagement signals: cancelled meetings, radio silence periods, hostile or dismissive tone, contradictory feedback, unexplained stakeholder changes. [integer]

**Mf** — Meeting frequency per week during active phases. [integer]

**Ma** — Client meeting attendance rate (attended meetings divided by scheduled meetings). [0,1]

**Es** — Escalation speed: how quickly the client raises blockers, concerns, or changing priorities to your team. [0,1]

**E_client** = client employee count

**E_ideal** = our ideal client size

<br>


Sz — Client company size fit. Uses a Gaussian function centred on our ideal client size to model the fact that clients too small (can't pay, unclear processes) or too large (bureaucratic, slow decisions) both present challenges:

Sz = e^( -(ln(E_client)-ln(E_deal))^2 / 2*σ^2 ) Where

## 6. Quality Assurance & Internal-External Review Cycle

This is the formal model of the dual-reviewer system. Every project has an internal reviewer (our team's quality gatekeeper) and an external reviewer (the client-side evaluator). This parameter captures the health of this feedback loop — whether deliverables are meeting standards, how aligned both reviewers are, whether defects are being resolved efficiently, and whether the review process itself is functioning. Variables I will be using are:

**Ir** — Internal reviewer score: the latest quality rating from your internal reviewer. [0,1]

**Er** — External reviewer (client-side) score: the latest quality rating from the client's designated reviewer. [0,1]

**Rc** — Total review cycles completed to date. [integer]

**Rp** — Review cycles planned for the entire project. [integer]

**Fr** — First-pass acceptance rate: fraction of deliverables approved on first submission without any revision requested. [0,1]

**Wt** — Weight given to the internal reviewer in the blended score. The external reviewer receives (1-Wt). [0,1]

## Now let's consider some binary hard knockout cases:

These are non-negotiable binary conditions. If any single gate fails, the project score drops to zero — the project should not be pursued, should be paused, or must be restructured before proceeding. No parameter score, however high, can compensate for a failed gate.

G1 — Contract or engagement authorisation. No signed contract, SOW, or formal engagement authorisation exists, and the client has no credible path to providing one. Work without a contract exposes the company to unrecoverable costs and legal liability.

G2 — Legal and ethical clearance. The project violates an NDA, non-compete, IP assignment, regulatory restriction, export control, or the company's ethical guidelines. This includes projects designed to deceive end users, facilitate fraud, or produce deliverables that cause harm.

G3 — Pipeline integrity. A duplicate or substantially identical project already exists in the pipeline or active delivery. Pursuing both creates internal conflict and resource waste.

G4 — Client entity verification. The client organisation is unverifiable, is a sanctioned entity, appears on a trade restriction list, or shows signals of being a shell company or front.

G5 — Data and IP ownership clarity. Ownership of all input data, deliverables, and generated IP is undefined, disputed, or contractually unacceptable — for example, the client demands ownership of your proprietary frameworks, reusable components, or internal tooling.

G6 — Minimum viable team. No qualified resource is assigned or available for this project. A project with zero allocated team members is a backlog entry, not a live project.

G7 — Client authorisation. The person authorising the project on the client side has no verified authority to commit budget, approve deliverables, or make binding decisions. If your contact lacks actual authority, the project is at risk of unilateral cancellation or repudiation.

G8 — Platform and infrastructure access. You do not have — and have no confirmed, timebound path to obtaining — all necessary system access, credentials, API keys, staging environments, hosting, and assets required to begin work. Without access, the team is blocked from day one.

G9 — Jurisdictional legality. The project requires deliverables in a jurisdiction where you legally cannot operate, or where the nature of the work is prohibited.

G10 — Conflict of interest. The project directly competes with an existing client under an exclusivity or non-compete arrangement.

G11 — Technical feasibility. After assessment, the project is determined to be fundamentally impossible with current or near-term available capabilities — not merely difficult, but infeasible.


They are all either 0 or 1, 0 when the above cases happen, instantly closing our deal or 1 if their respective cases don't happen.


G = G1 * G2 * G3 * G4 * G5 * G6 * G7 * G8 * G9 * G10 * G11


```python
G=1

for i in range(len(gates)):
  if gates[i]=='t':
    G=0
print(G)
```

## Situational Modifiers:

These are continuous multipliers — not binary gates — that capture real-world conditions which systematically affect project execution but do not fit cleanly into any single parameter. They adjust the final score up or down based on contextual factors.

**M1** — Data completeness confidence. Measures how much of the scoring data is based on confirmed information versus estimates, assumptions, or default values.

M1 = 0.5 + 0.5 * d_c

Where d_c is the proportion of all variables across all parameters that are populated from confirmed data rather than estimates, ranging from 0 to 1. The floor of 0.5 ensures that a project with entirely estimated data is not zeroed out, but it is significantly penalised.

**M2** — Organisational change risk. Measures whether the client is undergoing a merger, acquisition, leadership change, reorganisation, or strategic pivot that could destabilise the project.

1.00 — Stable organisation with no known changes

0.85 — Minor restructuring or leadership transition in a non-relevant department

0.70 — Major leadership or strategy change in progress that directly affects the project stakeholders

**M3** — Contract and commercial complexity. Measures how difficult the legal and commercial terms are to execute against.

1.00 — Standard terms and template SOW

0.85 — Custom terms required, some negotiation needed

0.70 — Complex legal negotiation expected, such as international or multi-entity arrangements

0.60 — Extreme contractual complexity: multiple jurisdictions, novel IP structures, heavy penalty clauses

**M4** — Currency and payment jurisdiction risk. Measures exposure to exchange rate fluctuations, cross-border payment delays, or clients in jurisdictions with payment reliability issues.

1.00 — Domestic, stable currency, reliable payment infrastructure

0.85 — International but stable market and well-established payment channels

0.70 — Emerging market, volatile currency, or delayed payment norms

**M5** — Execution complexity multiplier. Captures logistical difficulty: multiple time zones, language barriers, multi-location delivery, complex approval chains, or distributed stakeholder groups.

1.00 — Single-vertical, single-location, single-timezone, straightforward delivery

0.85 — Two verticals or two time zones

0.70 — Multi-vertical, multi-timezone

0.60 — Multi-vertical, multi-location, multi-language, complex stakeholder matrix

**M6** — Seasonal and cyclical pressure. Measures whether the project coincides with high-demand periods that compress timelines or compete for resources.

1.00 — Slow season: capacity is available, winning work is valuable, slight score boost

0.95 — Normal period

0.90 — Moderate pressure such as festival season or year-end

0.85 — Peak pressure period: multiple deadlines converging, team at capacity

**M7** — Sales and BD bandwidth. Whether you have the bandwidth to properly manage the opportunity through pre-sales, proposal, and negotiation.

1.00 — Dedicated team available for pitch and negotiation

0.85 — Available but stretched across multiple active deals

0.70 — Significantly stretched

0.50 — No one available to manage the opportunity

The combined modifier product is:

**M** = M1 * M2 * M3 * M4 * M5 * M6 * M7

```python
M=1

for i in range(len(modifiers)):
  M=M*modifiers[i]

print(M)
```
